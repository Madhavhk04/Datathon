```python
import pandas as pd
import numpy as np
import re

def clean_mixed_date(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    # Missing indicators
    if value.lower() in ["not available", "na", "n/a", "none", "null", ""]:
        return pd.NaT

    # Unix timestamp (10-digit seconds)
    if re.fullmatch(r"\d{10}", value):
        try:
            return pd.to_datetime(int(value), unit="s")
        except:
            return pd.NaT

    # YYYY-MM-DD / YYYY-MM-DD HH:MM:SS / ISO format
    if re.match(r"^\d{4}-\d{2}-\d{2}", value):
        try:
            return pd.to_datetime(value, dayfirst=False)
        except:
            return pd.NaT

    # YYYY/MM/DD
    if re.match(r"^\d{4}/\d{2}/\d{2}", value):
        try:
            return pd.to_datetime(value, dayfirst=False)
        except:
            return pd.NaT

    # DD/MM/YYYY
    if re.match(r"^\d{2}/\d{2}/\d{4}", value):
        try:
            return pd.to_datetime(value, dayfirst=True)
        except:
            return pd.NaT

    # DD-Mon-YYYY
    if re.match(r"^\d{1,2}-[A-Za-z]{3}-\d{4}", value):
        try:
            return pd.to_datetime(value, dayfirst=True)
        except:
            return pd.NaT

    # MM-DD-YYYY with AM/PM
    if re.match(r"^\d{2}-\d{2}-\d{4}.*(?:AM|PM)$", value, re.IGNORECASE):
        try:
            return pd.to_datetime(value, dayfirst=False)
        except:
            return pd.NaT

    # Anything else — try normal parsing without forcing dayfirst
    try:
        return pd.to_datetime(value, dayfirst=False)
    except:
        return pd.NaT
```

```python
df["hire_date_clean"] = df["hire_date"].apply(clean_mixed_date)
df["termination_date_clean"] = df["termination_date"].apply(clean_mixed_date)
```

```python
def real_parse_failures(original, cleaned):
    valid_original = (
        original.notna() &
        ~original.astype(str).str.strip().str.lower().isin(
            ["not available", "na", "n/a", "none", "null", ""]
        )
    )

    return (valid_original & cleaned.isna()).sum()

print(
    "Hire date real parse failures:",
    real_parse_failures(df["hire_date"], df["hire_date_clean"])
)

print(
    "Termination date real parse failures:",
    real_parse_failures(df["termination_date"], df["termination_date_clean"])
)
```

For joining firewall and endpoint data,  recommended removing .corp.local because the underlying hostname identifier is the WS/VDR/LPT + number.

```python
#hostname cleaning
import re

def clean_hostname(value):
    if pd.isna(value):
        return pd.NA

    value = str(value).strip().upper()

    # Remove corporate domain suffix
    value = re.sub(r"\.CORP\.LOCAL$", "", value)

    # Standardize underscore to hyphen
    value = value.replace("_", "-")

    return value

df["hostname_clean"] = df["hostname"].apply(clean_hostname)
```

```python
print(df[["hostname", "hostname_clean"]].dropna().sample(20, random_state=42))
```

```python
print("Missing hostnames:", df["hostname_clean"].isna().sum())
print("Unique cleaned hostnames:", df["hostname_clean"].nunique())
```

```python
hostname_check = (
    df[df["hostname_clean"].notna()]
    .groupby("hostname_clean")["user_id_clean"]
    .nunique()
)

print(
    "Hostnames mapped to multiple users:",
    (hostname_check > 1).sum()
)
```

```python
def clean_device_id(value):
    if pd.isna(value):
        return pd.NA

    value = str(value).strip().upper()

    # Remove spaces and hyphens
    value = re.sub(r"[\s-]+", "", value)

    return value

df["device_id_clean"] = df["device_id"].apply(clean_device_id)
```

```python
device_collisions = (
    df[df["device_id_clean"].notna()]
    .groupby("device_id_clean")["user_id_clean"]
    .nunique()
)

collision_ids = device_collisions[device_collisions > 1].index

df[df["device_id_clean"].isin(collision_ids)][
    ["user_id_clean", "username_clean", "device_id", "device_id_clean",
     "hostname_clean", "department_clean", "status_clean"]
].sort_values("device_id_clean")
```

```python
device_user_counts = (
    df.groupby("device_id_clean")["user_id_clean"]
    .transform("nunique")
)

df["device_id_conflict"] = (
    df["device_id_clean"].notna() &
    (device_user_counts > 1)
)
```

```python
print(df["device_id_conflict"].value_counts())
```

This isn't merely something to hide during cleaning. Shared/conflicting device assignments can themselves be a security signal.

For example, the data contains devices associated with employees from different departments and even different statuses. DEV25765, for instance, appears against three employees, including disabled and inactive users So our rule is:

Normalize → detect conflict → flag → preserve.

Not:

Normalize → delete records.
